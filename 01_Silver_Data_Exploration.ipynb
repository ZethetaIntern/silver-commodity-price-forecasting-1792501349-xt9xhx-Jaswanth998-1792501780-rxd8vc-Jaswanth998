```python
# Project 1D: Silver Technical Indicators

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Use the cleaned daily dataset
technical = daily_clean.copy()

# Sort records chronologically
technical["Date"] = pd.to_datetime(technical["Date"])
technical = technical.sort_values("Date").reset_index(drop=True)

# Calculate moving averages
technical["SMA_20"] = technical["Close"].rolling(20).mean()
technical["SMA_50"] = technical["Close"].rolling(50).mean()

# Calculate exponential moving average
technical["EMA_20"] = technical["Close"].ewm(
    span=20,
    adjust=False
).mean()

# Calculate RSI using 14 periods
delta = technical["Close"].diff()

gain = delta.clip(lower=0)
loss = -delta.clip(upper=0)

avg_gain = gain.ewm(
    alpha=1/14,
    min_periods=14,
    adjust=False
).mean()

avg_loss = loss.ewm(
    alpha=1/14,
    min_periods=14,
    adjust=False
).mean()

relative_strength = avg_gain / avg_loss

technical["RSI_14"] = (
    100 - (100 / (1 + relative_strength))
)

# Display the latest indicator values
display(
    technical[
        ["Date", "Close", "SMA_20", "SMA_50", "EMA_20", "RSI_14"]
    ].tail(10)
)

print("Technical indicators calculated successfully!")
```

```python
from pathlib import Path
import pandas as pd

# Project directory
PROJECT_DIR = Path(
    r"E:\Data Analyst\ZeTheta Projects\Project 1D Data Analyst Commodity Price Forecasting Agent"
)

# Cleaned data directory
CLEANED_DIR = PROJECT_DIR / "03_Cleaned_Data"

# Find the cleaned silver-price dataset
print("Searching for cleaned silver-price datasets...")

for file in CLEANED_DIR.glob("*.csv"):
    print(file.name)
```

```python
# Load the cleaned daily silver-price dataset

silver_file = CLEANED_DIR / "silver_daily_ohlcv_cleaned.csv"

silver_df = pd.read_csv(silver_file)

# Convert Date to datetime
silver_df["Date"] = pd.to_datetime(silver_df["Date"], errors="coerce")

# Sort data chronologically
silver_df = silver_df.sort_values("Date").reset_index(drop=True)

# Display dataset information
print("Dataset loaded successfully!")
print("Number of rows:", silver_df.shape[0])
print("Number of columns:", silver_df.shape[1])

print("\nColumn names:")
print(silver_df.columns.tolist())

print("\nFirst 5 rows:")
display(silver_df.head())

print("\nData types:")
print(silver_df.dtypes)
```

```python
# Step 3: Create machine-learning features

import pandas as pd
import numpy as np

# Work with a copy of the dataset
ml_df = silver_df.copy()

# Ensure chronological order
ml_df = ml_df.sort_values("Date").reset_index(drop=True)

# Create historical price features
ml_df["Close_Lag_1"] = ml_df["Close"].shift(1)
ml_df["Close_Lag_2"] = ml_df["Close"].shift(2)
ml_df["Close_Lag_3"] = ml_df["Close"].shift(3)
ml_df["Close_Lag_5"] = ml_df["Close"].shift(5)
ml_df["Close_Lag_10"] = ml_df["Close"].shift(10)

# Create rolling averages using past prices only
ml_df["SMA_5"] = ml_df["Close"].shift(1).rolling(5).mean()
ml_df["SMA_10"] = ml_df["Close"].shift(1).rolling(10).mean()
ml_df["SMA_20"] = ml_df["Close"].shift(1).rolling(20).mean()

# Target: next trading day's closing price
ml_df["Target_Next_Close"] = ml_df["Close"].shift(-1)

# Remove rows without sufficient historical data or a target
ml_df = ml_df.dropna().reset_index(drop=True)

print("Forecasting features created successfully!")
print("Rows:", ml_df.shape[0])
print("Columns:", ml_df.shape[1])

print("\nNew feature columns:")
print([
    "Close_Lag_1",
    "Close_Lag_2",
    "Close_Lag_3",
    "Close_Lag_5",
    "Close_Lag_10",
    "SMA_5",
    "SMA_10",
    "SMA_20",
    "Target_Next_Close"
])

display(ml_df[
    ["Date", "Close", "Close_Lag_1", "SMA_5", "SMA_20",
     "Target_Next_Close"]
].head())
```

```python
# Step 4: Prepare training and testing datasets

from sklearn.model_selection import train_test_split

# Select the forecasting features
FEATURES = [
    "Close_Lag_1",
    "Close_Lag_2",
    "Close_Lag_3",
    "Close_Lag_5",
    "Close_Lag_10",
    "SMA_5",
    "SMA_10",
    "SMA_20"
]

TARGET = "Target_Next_Close"

# Create feature matrix and target
X = ml_df[FEATURES]
y = ml_df[TARGET]

# Chronological 80/20 split (no shuffling)
split_index = int(len(ml_df) * 0.80)

X_train = X.iloc[:split_index].copy()
X_test = X.iloc[split_index:].copy()

y_train = y.iloc[:split_index].copy()
y_test = y.iloc[split_index:].copy()

print("Training and testing split completed!")
print("Training rows:", len(X_train))
print("Testing rows:", len(X_test))

print("\nTraining period:")
print(ml_df["Date"].iloc[0], "to",
      ml_df["Date"].iloc[split_index - 1])

print("\nTesting period:")
print(ml_df["Date"].iloc[split_index], "to",
      ml_df["Date"].iloc[-1])

print("\nNumber of features:", len(FEATURES))
```

```python
# Step 5: Train and compare machine learning models

import pandas as pd
import numpy as np

from sklearn.ensemble import RandomForestRegressor, ExtraTreesRegressor
from sklearn.linear_model import LinearRegression
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

# Initialize models
models = {
    "Linear Regression": LinearRegression(),
    "Random Forest": RandomForestRegressor(
        n_estimators=100,
        random_state=42,
        n_jobs=-1
    ),
    "Extra Trees": ExtraTreesRegressor(
        n_estimators=100,
        random_state=42,
        n_jobs=-1
    )
}

results = []
trained_models = {}

# Train and evaluate each model
for name, model in models.items():

    model.fit(X_train, y_train)

    predictions = model.predict(X_test)

    mae = mean_absolute_error(y_test, predictions)
    rmse = np.sqrt(mean_squared_error(y_test, predictions))
    r2 = r2_score(y_test, predictions)

    results.append({
        "Model": name,
        "MAE": mae,
        "RMSE": rmse,
        "R2 Score": r2
    })

    trained_models[name] = model

# Display comparison
results_df = pd.DataFrame(results)
results_df = results_df.sort_values("MAE")

display(results_df)

print("Model training and evaluation completed!")
```

```python
# Step 6: Actual vs Predicted Silver Prices

import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Select the best-performing model
best_model_name = results_df.iloc[0]["Model"]
best_model = trained_models[best_model_name]

# Generate predictions
y_pred = best_model.predict(X_test)

# Prepare comparison
comparison_df = pd.DataFrame({
    "Actual_Price": np.asarray(y_test),
    "Predicted_Price": y_pred
})

# Plot actual and predicted prices
plt.figure(figsize=(15, 6))

plt.plot(
    comparison_df["Actual_Price"].values,
    label="Actual Silver Price",
    linewidth=1.5
)

plt.plot(
    comparison_df["Predicted_Price"].values,
    label="Predicted Silver Price",
    linewidth=1.5
)

plt.title(f"Actual vs Predicted Silver Prices - {best_model_name}")
plt.xlabel("Test Observation")
plt.ylabel("Silver Price")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()

# Save chart
chart_path = PROJECT_DIR / "actual_vs_predicted_silver_prices.png"
plt.savefig(chart_path, dpi=300, bbox_inches="tight")
plt.show()

# Save predictions
prediction_path = PROJECT_DIR / "silver_price_predictions.csv"
comparison_df.to_csv(prediction_path, index=False)

print("Best model:", best_model_name)
print("Chart saved to:", chart_path)
print("Predictions saved to:", prediction_path)

display(comparison_df.head(10))
```

```python
# Step 9B: Generate 30-day recursive silver-price forecasts

import pandas as pd
import numpy as np
from pathlib import Path

# Sort historical data by date
history = ml_df[["Date", "Close"]].copy()
history["Date"] = pd.to_datetime(history["Date"])
history = history.sort_values("Date").reset_index(drop=True)

# Remove invalid closing prices
history = history.dropna(subset=["Date", "Close"])
history = history[history["Close"] > 0]

# Prepare historical closing prices
prices = history["Close"].astype(float).tolist()
last_date = history["Date"].iloc[-1]

# Select the saved best model
forecast_model = joblib.load(model_path)

# Use the exact features used during training
print("Model features:", FEATURES)
print("Last historical date:", last_date.date())
print("Last historical close:", prices[-1])

# Verify that the features can be reconstructed
required_features = [
    "Close_Lag_1",
    "Close_Lag_2",
    "Close_Lag_3",
    "Close_Lag_5",
    "Close_Lag_10",
    "SMA_5",
    "SMA_10",
    "SMA_20"
]

if set(FEATURES) != set(required_features):
    raise ValueError(
        "The model features differ from the expected lag and moving-average "
        "features. Inspect FEATURES before forecasting."
    )

# Generate forecasts recursively
forecast_rows = []
forecast_prices = prices.copy()
forecast_date = last_date

for step in range(1, 31):

    # Construct features using available historical/predicted prices
    feature_values = {
        "Close_Lag_1": forecast_prices[-1],
        "Close_Lag_2": forecast_prices[-2],
        "Close_Lag_3": forecast_prices[-3],
        "Close_Lag_5": forecast_prices[-5],
        "Close_Lag_10": forecast_prices[-10],
        "SMA_5": np.mean(forecast_prices[-5:]),
        "SMA_10": np.mean(forecast_prices[-10:]),
        "SMA_20": np.mean(forecast_prices[-20:])
    }

    X_future = pd.DataFrame(
        [[feature_values[col] for col in FEATURES]],
        columns=FEATURES
    )

    predicted_close = float(forecast_model.predict(X_future)[0])

    if not np.isfinite(predicted_close) or predicted_close <= 0:
        raise ValueError(
            f"Invalid forecast generated at step {step}: {predicted_close}"
        )

    # Move to the next calendar date.
    # This is a calendar-day forecast, not a trading-calendar forecast.
    forecast_date += pd.Timedelta(days=1)

    forecast_rows.append({
        "Forecast_Step": step,
        "Forecast_Date": forecast_date,
        "Predicted_Close": predicted_close
    })

    # Use the prediction to generate the next forecast
    forecast_prices.append(predicted_close)

forecast_df = pd.DataFrame(forecast_rows)

# Save forecast results
forecast_dir = PROJECT_DIR / "06_Forecasts"
forecast_dir.mkdir(parents=True, exist_ok=True)

forecast_file = forecast_dir / "silver_price_30_day_forecast.csv"
forecast_df.to_csv(forecast_file, index=False)

print("\nForecast generated successfully!")
print("Forecast rows:", len(forecast_df))
print("Saved to:", forecast_file)

display(forecast_df.head(10))
```

```python
# Step 12: Compare All Models Against the Baseline

import pandas as pd
import numpy as np
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Step 1: Prepare actual test prices
actual = np.asarray(y_test).ravel()

# Step 2: Generate baseline predictions
# Previous closing price predicts the next closing price
baseline_predictions = X_test["Close_Lag_1"].to_numpy()

# Step 3: Initialize results
results = []

# Step 4: Evaluate the baseline model
baseline_mae = mean_absolute_error(actual, baseline_predictions)
baseline_rmse = np.sqrt(mean_squared_error(actual, baseline_predictions))
baseline_r2 = r2_score(actual, baseline_predictions)

results.append({
    "Model": "Baseline",
    "MAE": baseline_mae,
    "RMSE": baseline_rmse,
    "R2": baseline_r2
})

# Step 5: Evaluate all trained models
for model_name, model in models.items():

    predictions = np.asarray(model.predict(X_test)).ravel()

    model_mae = mean_absolute_error(actual, predictions)
    model_rmse = np.sqrt(mean_squared_error(actual, predictions))
    model_r2 = r2_score(actual, predictions)

    results.append({
        "Model": model_name,
        "MAE": model_mae,
        "RMSE": model_rmse,
        "R2": model_r2
    })

# Step 6: Create the comparison table
comparison = pd.DataFrame(results)

# Lower MAE is better
comparison = comparison.sort_values(
    by="MAE",
    ascending=True
).reset_index(drop=True)

# Step 7: Display comparison results
print("=" * 65)
print("SILVER PRICE FORECASTING - MODEL COMPARISON")
print("=" * 65)

display(comparison.round(4))

# Step 8: Identify the best model
best_model = comparison.iloc[0]["Model"]

print("\nBest model by MAE:", best_model)

# Step 9: Compare trained models against the baseline
print("\nBASELINE COMPARISON")
print("-" * 65)

for _, row in comparison.iterrows():

    if row["Model"] == "Baseline":
        continue

    mae_difference = row["MAE"] - baseline_mae

    if mae_difference < 0:
        improvement = (
            (baseline_mae - row["MAE"]) / baseline_mae
        ) * 100

        print(
            f"{row['Model']}: MAE improved by "
            f"{improvement:.2f}% compared with the baseline."
        )

    elif mae_difference > 0:
        difference = (
            (row["MAE"] - baseline_mae) / baseline_mae
        ) * 100

        print(
            f"{row['Model']}: MAE is "
            f"{difference:.2f}% higher than the baseline."
        )

    else:
        print(
            f"{row['Model']}: MAE matches the baseline."
        )

# Step 10: Final evaluation summary
best_mae = comparison.iloc[0]["MAE"]

print("\nFINAL EVALUATION")
print("-" * 65)

if best_model == "Baseline":
    print(
        "The baseline achieved the lowest MAE. "
        "The trained models did not outperform it on MAE."
    )
else:
    print(
        f"{best_model} achieved the lowest MAE "
        "among the evaluated models."
    )

print(f"Baseline MAE: {baseline_mae:.4f}")
print(f"Best MAE: {best_mae:.4f}")
print("\nModel comparison completed successfully!")
```
